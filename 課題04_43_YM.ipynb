# **神経衰弱**



## **概要**

このプログラムは、Pythonを使って作成した神経衰弱ゲームです．プレイヤーはカードの中から2枚を選び、同じカードのペアを見つけていきます．ゲームを開始すると、最初に難易度を選択します．難易度はEASY、NORMAL、HARDの3種類があり、EASYでは4×4の16枚、NORMALでは4×6の24枚、HARDでは6×6の36枚のカードを使用します．難易度が高くなるほどカードの枚数が増えるため、より多くのペアを見つける必要があります．
カードはゲーム開始時にランダムに並べられるため、毎回違う配置で遊ぶことができます．プレイヤーは表示された番号を入力して2枚のカードを選択します．2枚が同じだった場合はそのまま残り、違っていた場合はカードが伏せられます．すべてのペアを見つけるとゲームクリアになります．
また、ターン数やミス回数、プレイ時間を記録し、ゲーム終了時にスコアを表示するようにしました．難易度によってスコアボーナスも変わるようにしているため、HARDで高得点を目指すこともできます．ゲーム終了後には、もう一度プレイするかどうかを選択できます．

## **操作方法**
プログラムを実行すると、最初に難易度を選択する画面が表示されます．

1 : EASY   (4×4)
2 : NORMAL (4×6)
3 : HARD   (6×6)

遊びたい難易度に対応した番号を入力してください．EASYでは16枚、NORMALでは24枚、HARDでは36枚のカードを使用します．
難易度を選択すると、カードがランダムに配置されます．盤面にはカードの位置を表す番号が表示され、最初はすべて「?」になっています．
ゲーム中は「1枚目のカード番号」と表示されるので、選びたいカードの番号を入力してください．その後、「2枚目のカード番号」と表示されるので、別のカードの番号を入力してください．
2枚のカードが同じ場合はペアとして残り、違う場合はカードが伏せられます．すでに揃っているカードや、1枚目に選んだカードをもう一度選ぶことはできません．
すべてのペアを揃えるとゲームクリアとなり、難易度、ターン数、ミス回数、プレイ時間、スコアが表示されます．
ゲーム終了後に「もう一度遊びますか？」と表示されます．「y」を入力すると新しいゲームが始まり、「n」を入力するとゲームが終了します．
入力するときは、表示されている番号を確認してから入力してください．数字以外を入力した場合や、選択できないカードを入力した場合は、もう一度入力するように表示されます．

## **制作者のコメント**
Pythonを使って実際に遊ぶことができるゲームを作ることを目標にしました．最初は神経衰弱をPythonで作るのは難しそうだと思いましたが、カードを作る処理、盤面を表示する処理、カードを選択する処理、ペアを判定する処理などに分けて考えることで、少しずつプログラムを作っていきました．
特に工夫した点は、最初に難易度を選択できるようにしたことです．EASY、NORMAL、HARDの3種類を用意し、難易度によってカードの枚数が変わるようにしました．また、難しいモードほどスコアにボーナスが入るようにしたため、ただクリアするだけではなく、高い難易度で高得点を狙う楽しみも作りました．

さらに、ターン数、ミス回数、プレイ時間を記録して、最後にスコアとして表示するようにしました.

```python
%reset -f
import random
import time


# カードを作る
def make_cards(pair_count):
    symbols = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
    cards = symbols[:pair_count] * 2
    random.shuffle(cards)
    return cards


# 盤面を表示する
def show_board(cards, opened, matched, columns):
    print()
    print("          MEMORY GAME")
    print("------------------------------")

    for i in range(len(cards)):
        if i in matched or i in opened:
            print(f" {cards[i]} ", end="")
        else:
            print(" ?  ", end="")

        if (i + 1) % columns == 0:
            print()

    print("------------------------------")

    for i in range(len(cards)):
        print(f"{i + 1:2}", end="   ")

        if (i + 1) % columns == 0:
            print()

    print()


# カードを選ぶ
def select_card(cards, opened, matched, message):
    while True:
        try:
            number = int(input(message)) - 1

            if number < 0 or number >= len(cards):
                print(f"1～{len(cards)}の番号を入力してください。")

            elif number in matched:
                print("そのカードはすでに揃っています。")

            elif number in opened:
                print("そのカードはすでに選択しています。")

            else:
                return number

        except ValueError:
            print("数字を入力してください。")


# スコア計算
def calculate_score(turns, misses, difficulty):
    score = 1000

    score -= turns * 10
    score -= misses * 20

    # 難易度によるボーナス
    if difficulty == "HARD":
        score += 500
    elif difficulty == "NORMAL":
        score += 250

    if score < 0:
        score = 0

    return score


# 難易度選択
def choose_difficulty():

    while True:

        print()
        print("==============================")
        print("       難易度を選択")
        print("==============================")
        print("1 : EASY   (4×4)")
        print("2 : NORMAL (4×6)")
        print("3 : HARD   (6×6)")
        print("==============================")

        choice = input("番号を入力してください → ")

        if choice == "1":
            return "EASY", 8, 4

        elif choice == "2":
            return "NORMAL", 12, 6

        elif choice == "3":
            return "HARD", 18, 6

        else:
            print("1～3のどれかを入力してください。")


# 神経衰弱本体
def memory_game():

    difficulty, pair_count, columns = choose_difficulty()

    cards = make_cards(pair_count)

    opened = []
    matched = []

    turns = 0
    misses = 0

    print()
    print("================================")
    print("          MEMORY GAME")
    print("================================")
    print(f"難易度 : {difficulty}")
    print(f"カード : {len(cards)}枚")
    print()
    print("同じカードを2枚ずつ揃えてください！")

    start_time = time.time()

    while len(matched) < len(cards):

        show_board(cards, opened, matched, columns)

        print(f"ターン数 : {turns}")
        print(f"ミス回数 : {misses}")
        print()

        # 1枚目
        first = select_card(
            cards,
            opened,
            matched,
            "1枚目のカード番号 → "
        )

        opened.append(first)

        show_board(cards, opened, matched, columns)

        print("1枚目:", cards[first])
        print()

        # 2枚目
        second = select_card(
            cards,
            opened,
            matched,
            "2枚目のカード番号 → "
        )

        opened.append(second)

        show_board(cards, opened, matched, columns)

        print("1枚目:", cards[first])
        print("2枚目:", cards[second])

        turns += 1

        # ペア判定
        if cards[first] == cards[second]:

            print()
            print("🎉 ペアが揃いました！")

            matched.append(first)
            matched.append(second)

            time.sleep(1)

        else:

            print()
            print("❌ ペアではありません。")

            misses += 1

            time.sleep(1)

        opened = []

        print()


    # ゲーム終了

    end_time = time.time()

    play_time = int(end_time - start_time)

    score = calculate_score(
        turns,
        misses,
        difficulty
    )

    show_board(cards, [], matched, columns)

    print()
    print("================================")
    print("          CLEAR!!")
    print("================================")
    print(f"難易度 : {difficulty}")
    print(f"ターン数 : {turns}")
    print(f"ミス回数 : {misses}")
    print(f"プレイ時間 : {play_time} 秒")
    print(f"スコア : {score} 点")
    print("================================")
    print()


# メイン処理

while True:

    memory_game()

    while True:

        again = input("もう一度遊びますか？ (y/n) → ").lower()

        if again == "y":
            print()
            break

        elif again == "n":
            print()
            print("プレイしてくれてありがとうございました！")
            print("ゲームを終了します。")
            break

        else:
            print("y または n を入力してください。")

    if again == "n":
        break
```
